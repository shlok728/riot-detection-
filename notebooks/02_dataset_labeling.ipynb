# 02 - Dataset Labeling

This notebook analyzes the original labels and metadata of the video datasets
used for the Crowd Violence / Riot Risk Detection project.

Objectives:
- Inspect the original labeling structure
- Analyze class distributions
- Understand video-to-label relationships
- Preserve dataset provenance
- Prepare a unified video-level manifest

Important:
- Original dataset files will not be modified.
- Dataset splits will not be created here.
- No frame extraction or model training will be performed here.

```python
from collections import Counter

vf_label_counts = Counter()

for video in vf_videos:
    parts = video.parts

    if "Violence" in parts:
        vf_label_counts["violence"] += 1
    elif "NonViolence" in parts:
        vf_label_counts["nonviolence"] += 1
    else:
        vf_label_counts["unknown"] += 1

print("Violent-Flows labels from directory structure:")
for label, count in vf_label_counts.items():
    print(f"{label}: {count}")
```

```python
print("Checking for videos with unknown labels...\n")

unknown_videos = [
    video for video in vf_videos
    if "Violence" not in video.parts and "NonViolence" not in video.parts
]

print("Unknown videos:", len(unknown_videos))

for video in unknown_videos:
    print(video)
```

```python
rwf_records = []

for video in RWF_DIR.rglob("*"):
    if not video.is_file() or video.suffix.lower() != ".avi":
        continue

    parts = video.parts

    if "Train_Fight" in parts:
        label = "violence"
        split = "train"
    elif "Train_NonFight" in parts:
        label = "nonviolence"
        split = "train"
    elif "Val_Fight" in parts:
        label = "violence"
        split = "val"
    elif "Val_NonFight" in parts:
        label = "nonviolence"
        split = "val"
    else:
        continue

    rwf_records.append({
        "video_path": str(video),
        "source_dataset": "RWF-2000",
        "source_video_id": video.stem,
        "original_split": split,
        "label": label
    })

print("RWF-2000 records:", len(rwf_records))
```

```python
vf_records = []

for video in vf_videos:
    parts = video.parts

    if "Violence" in parts:
        label = "violence"
    elif "NonViolence" in parts:
        label = "nonviolence"
    else:
        continue

    # The numbered folder (1–5) preserves the original dataset grouping.
    group = next(
        (part for part in parts if part in {"1", "2", "3", "4", "5"}),
        None
    )

    vf_records.append({
        "video_path": str(video),
        "source_dataset": "Violent-Flows",
        "source_video_id": video.stem,
        "original_split": group,
        "label": label
    })

print("Violent-Flows records:", len(vf_records))
```

```python
all_records = rwf_records + vf_records

raw_manifest = pd.DataFrame(all_records)

print("Total records:", len(raw_manifest))
print("\nClass distribution:")
print(raw_manifest["label"].value_counts())

print("\nDataset distribution:")
print(raw_manifest["source_dataset"].value_counts())

display(raw_manifest.head())
```

```python
print("Missing values:")
print(raw_manifest.isnull().sum())

print("\nDuplicate rows:", raw_manifest.duplicated().sum())

print(
    "Duplicate video paths:",
    raw_manifest["video_path"].duplicated().sum()
)
```

```python
print("\nRecords by dataset and label:")
print(
    raw_manifest
    .groupby(["source_dataset", "label"])
    .size()
)
```
